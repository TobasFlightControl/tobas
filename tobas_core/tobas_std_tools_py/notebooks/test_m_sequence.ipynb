```python
import numpy as np
from matplotlib import pyplot as plt
from scipy import signal, fft

from tobas_std_tools_py.control.system_identification.m_sequence_generator import MSequenceGenerator
```

```python
# M系列信号を生成

n = 9

m_sequence_gen = MSequenceGenerator(n)
m_sequence = [bit for bit in m_sequence_gen]
m_sequence = np.array(m_sequence)

print(f'信号の平均値: {np.mean(m_sequence)}')
print(f'理論値: {1 / m_sequence_gen.length()}')
```

```python
# 波形を描画
# システム同定の基礎: 図5.5, (a)

plt.plot(m_sequence)
plt.title('M Sequence Waveform')
plt.show()
```

```python
# 自己相関関数を描画
# システム同定の基礎: 図5.5, (b)

# scipy.signal.correlateは，二つのシグナル間の相互相関（cross-correlation）を計算する．相互相関は，一方のシグナルが他方のシグナルにどの程度似ているかを異なるタイムラグで計算したものである．
# 同じシグナルに対する相互相関は自己相関（autocorrelation）と呼ばれ，シグナルの周期性を解析するのに用いられる．
autocorr = signal.correlate(m_sequence, m_sequence) / m_sequence_gen.length()
autocorr = autocorr[:m_sequence_gen.length()]

plt.plot(autocorr)
plt.title("Autocorrelation Function")
plt.xlabel("Lag")
plt.show()
```

```python
# パワースペクトルを描画
# システム同定の基礎: 図5.5, (c)

# FFTを計算
# インデックスはその周波数成分がどれだけの頻度で現れるか（つまりその周波数 [Hz]）を表している
# 各要素は特定の周波数成分の振幅と位相を表す複素数で，実部はコサイン波の振幅を，虚部はサイン波の振幅を表す．
m_sequence_fft = fft.fft(m_sequence)

# FFTの結果を整形
m_sequence_fft = m_sequence_fft[:len(m_sequence_fft) // 2]  # 複素共役を削除
spectrum = np.abs(m_sequence_fft)**2  # パワースペクトルを計算
spectrum_db = 10 * np.log10(spectrum)  # dBスケールに変換

# 規格化周波数
f_norm = np.linspace(0, 1, len(spectrum_db))

plt.plot(f_norm, spectrum_db)
plt.title("Power Spectrum [dB]")
plt.xlabel("Normalized Frequency")
plt.ylabel("Power Spectrum [dB]")
plt.show()
```
