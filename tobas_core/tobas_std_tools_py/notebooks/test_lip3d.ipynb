```python
import math
import numpy as np
import matplotlib as mpl
from matplotlib import pyplot as plt
from matplotlib import patches

from tobas_std_tools_py.control.humanoid import lip3d
```

```python
def plot_trajectory(s, T_sup, zc, fp0, p0, v0, r, footsize=0.05):
    # 軌道生成
    n = len(s) + 1
    traj, fp_des_list, fp_list = lip3d(s, T_sup, zc, fp0, p0, v0, r)

    fig = plt.figure(figsize=(20, 20))
    ax = fig.add_subplot(1, 1, 1)

    # 重心軌道
    cx = []
    cy = []
    for t in np.linspace(0., T_sup * n, 1000):
        p, v = traj(t)
        cx.append(p[0])
        cy.append(p[1])
    ax.plot(cx, cy, c='k')

    # 目標接地点と実際の値
    px_des, py_des = np.hsplit(fp_des_list, 2)
    px, py = np.hsplit(fp_list, 2)
    ax.scatter(px_des, py_des, c='k', marker='x', label='desired')
    ax.scatter(px, py, c='k', marker='o', label='modified')

    # 歩行素片の変換点
    change_x = []
    change_y = []
    for i in range(n):
        t = T_sup * i
        p, v = traj(t)
        change_x.append(p[0])
        change_y.append(p[1])
    ax.scatter(change_x, change_y, c='k', marker='^')

    # 足跡
    for i in range(n):
        patch = patches.RegularPolygon(
            xy=(px_des[i], py_des[i]),
            numVertices=4,
            radius=footsize,
            orientation=(0. if i == 0 else s[i - 1][2]) - np.pi / 4.,
            fill=False,
            linestyle=':',
        )
        ax.add_patch(patch)
        patch = patches.RegularPolygon(
            xy=(px[i], py[i]),
            numVertices=4,
            radius=footsize,
            orientation=(0. if i == 0 else s[i - 1][2]) - np.pi / 4.,
            fill=False,
            linestyle='-',
        )
        ax.add_patch(patch)

    ax.set_aspect('equal')
    ax.set_xlim(min(min(px), min(px_des)) - footsize * 1.5, max(max(px), max(px_des)) + footsize * 1.5)
    ax.set_ylim(min(min(py), min(py_des)) - footsize * 1.5, max(max(py), max(py_des)) + footsize * 1.5)
    plt.legend()
    plt.xlabel('x')
    plt.ylabel('y')
    plt.show()
```

```python
# 図4.24

s = [
    (0., 0.2, 0.),
    (0.3, 0.2, 0.),
    (0.3, 0.2, 0.),
    (0.3, 0.2, 0.),
    (0., 0.2, 0.),
]
T_sup = 0.8
zc = 0.8
fp0 = (0., 0.)      # 足場(foot pos)の初期値
p0 = (0., 0.01)     # CoMが初期足位置の鉛直線上にあると上手く軌道が作れない
v0 = (0., 0.)
r = 10.

plot_trajectory(s, T_sup, zc, fp0, p0, v0, r)
```

```python
# 図4.25

s = [
    (0., 0.2, 0.),
    (0.2, 0.3, 0.),
    (0.2, 0.1, 0.),
    (0.2, 0.3, 0.),
    (0., 0.2, 0.),
]
T_sup = 0.8
zc = 0.8
fp0 = (0., 0.)
p0 = (0., 0.01)
v0 = (0., 0.)
r = 10.

plot_trajectory(s, T_sup, zc, fp0, p0, v0, r)
```

```python
# 図4.27

s = [
    (0., 0.2, math.radians(0.)),
    (0.25, 0.2, math.radians(20.)),
    (0.25, 0.2, math.radians(40.)),
    (0.25, 0.2, math.radians(60.)),
    (0., 0.2, math.radians(60.)),
]
T_sup = 0.8
zc = 0.8
fp0 = (0., 0.)
p0 = (0., 0.01)
v0 = (0., 0.)
r = 10.

plot_trajectory(s, T_sup, zc, fp0, p0, v0, r)
```
